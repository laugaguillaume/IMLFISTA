```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Style propre et lisible
sns.set_theme()
sns.color_palette("pastel")

# Données factices
n_iter = 100
loss_cond = np.exp(-0.05 * np.arange(n_iter)) + 0.05 * np.random.rand(n_iter)
loss_mlfb = np.exp(-0.04 * np.arange(n_iter)) + 0.05 * np.random.rand(n_iter)
loss_fb = np.exp(-0.03 * np.arange(n_iter)) + 0.05 * np.random.rand(n_iter)

# Quelques cycles arbitraires
cycles_cond = [10, 30, 60, 90]
cycles_mlfb = [20, 50, 80]

# Tracé
plt.figure(figsize=(7, 5))
plt.plot(loss_cond, label="BCD cond", linewidth=2)
plt.plot(loss_mlfb, label="BCD MLFB", linewidth=2)
plt.plot(loss_fb, label="FB", linewidth=2)

# Points pour cycles_cond
plt.scatter(cycles_cond, [loss_cond[i] for i in cycles_cond], marker="o", s=80, label="cycles_cond")

# Points pour cycles_mlfb
plt.scatter(cycles_mlfb, [loss_mlfb[i] for i in cycles_mlfb], marker="x", s=80, label="cycles_mlfb")

# Habillage
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.title("Loss over Iterations")
plt.legend(frameon=True)
plt.tight_layout()

plt.show()
```
